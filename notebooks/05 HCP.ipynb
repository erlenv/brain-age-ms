# HCP Young Adults: Model Validation

This notebook validates the three brain age models on the HCP Young Adult
dataset, which serves as a healthy control benchmark for the OFAMS analyses.

Since the HCP cohort consists of healthy adults with a narrow age range
(22--37 years), it provides a reference for evaluating how well each model
predicts chronological age in the absence of pathology. Performance metrics
computed here are used as a baseline for interpreting model accuracy in the
clinical OFAMS cohort.

## Contents

**Setup**
Imports, plot style, and model color conventions. Prediction files from
all three models are loaded alongside HCP metadata, merged on subject ID,
and column names are standardized to match the conventions used across
all notebooks.

**BAG computation**
Brain Age Gap (BAG = predicted age − chronological age) is computed for
each subject and model.

**Model accuracy**
MAE, RMSE, Pearson r, and Spearman ρ between predicted and chronological
age are reported for each model, providing a direct comparison against
benchmarks from the original model publications.

**Visualisation**
Scatter plots of predicted age vs. chronological age and BAG vs.
chronological age are generated for each model, allowing visual inspection
of prediction quality and age bias.

```python
# Imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
from scipy import stats
from sklearn.metrics import mean_absolute_error

sns.set_theme(style="whitegrid")
```

## Setup

```python
# Load prediction data and metadata
brainager = pd.read_csv("brainageR_hcp.csv")
pyment = pd.read_csv("pyment_hcp.csv")
hba = pd.read_csv("hba_hcp.csv")
metadata = pd.read_csv("HCP_metadata.csv")
hba
```

### Standardize columns and merge data

```python
#HBA
# Rename cols
hba = hba.rename(columns={"eid": "Subject", "brainage":"predicted_age", "age":"Age_in_Yrs"})
hba = hba[["Subject","predicted_age","Age_in_Yrs"]]

# Merge with HCP metadata on subject
hba = metadata[["Subject", "Age_in_Yrs","Gender"]].merge(hba[["Subject", "predicted_age", "Age_in_Yrs"]],                                                          on=["Subject", "Age_in_Yrs"], how="left")
hba
```

```python
#pyment
# Rename cols
pyment = pyment.rename(columns={"subject_id": "Subject"})

#Merge with HCP metadata on subject
pyment = metadata[["Subject", "Age_in_Yrs","Gender"]].merge(pyment[["Subject", "predicted_age"]],
    on="Subject",
    how="left"
)
pyment
```

```python
#brainageR
#Rename cols
brainager = brainager.rename(columns={"subject_id": "Subject", "brain.predicted_age": "predicted_age"})

# Merge with HCP metadata on subject
brainager = metadata[["Subject", "Age_in_Yrs", "Gender"]].merge(
    brainager[["Subject", "predicted_age"]],
    on="Subject",
    how="left"
)
```

```python
brainager
```

## BAG computation
BAG = predicted age - chronological age

```python
brainager["BAG"] = brainager["predicted_age"] - brainager["Age_in_Yrs"]
pyment["BAG"] = pyment["predicted_age"] - pyment["Age_in_Yrs"]
hba["BAG"] = hba["predicted_age"] - hba["Age_in_Yrs"]
```

## Model accuracy

```python
# Define model dict
models = {
    "brainageR": brainager,
    "pyment": pyment,
    "HBA": hba
}

# Calculate accuracy scores for each model
for name, df in models.items():
    y_true = df["Age_in_Yrs"]
    y_pred = df["predicted_age"]
    mae  = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(np.mean((y_pred - y_true) ** 2))
    r, p_r   = stats.pearsonr(y_true, y_pred)
    rho, p_s = stats.spearmanr(y_true, y_pred)

    print(f"\n── {name} validation (N={len(df)} HCP YA) ──")
    print(f"MAE:        {mae:.2f} years")
    print(f"RMSE:       {rmse:.2f} years")
    print(f"Pearson r:  {r:.3f}  (p={p_r:.4f})")
    print(f"Spearman ρ: {rho:.3f}  (p={p_s:.4f})")
    print(f"BAG:        mean={df['BAG'].mean():.2f}, SD={df['BAG'].std():.2f}")
```

## Visualisations

### Predicted age vs. chronological age

```python
# Define model dict
models = {
    "pyment": pyment,
    "brainageR": brainager,
    "HBA": hba
}

# Define model colours
palette_models = {
    "pyment":    "#7B4FA6",
    "brainageR": "#3A8A4A",
    "HBA":       "#D94F4F",
}

# Define axe limits
age_min, age_max = 10, 55
x_line = np.array([age_min, age_max])

# Create figure
fig = plt.figure(figsize=(18, 14))

axes = [
    fig.add_subplot(2, 2, 1),
    fig.add_subplot(2, 2, 2),
    fig.add_subplot(2, 2, (3, 4))
]

# Center third subplot
axes[2].set_position([0.25, 0.05, 0.5, 0.35])

fig.suptitle("Predicted age versus chronological age - HCP",
             fontsize=28, fontweight="bold", y=1.01)

for ax, (name, df) in zip(axes, models.items()):
    mae  = mean_absolute_error(df["Age_in_Yrs"], df["predicted_age"])
    rmse = np.sqrt(np.mean((df["predicted_age"] - df["Age_in_Yrs"]) ** 2))

    sns.scatterplot(data=df, x="Age_in_Yrs", y="predicted_age",
                    color=palette_models[name], alpha=0.8,
                    edgecolor="white", linewidth=0.5, s=100, ax=ax)

    ax.plot([age_min, age_max], [age_min, age_max], "k--", lw=1.5,
            label="Identity (y = x)")

    ax.set_xlim(age_min, age_max)
    ax.set_ylim(age_min, age_max)
    ax.set_xlabel("Chronological age (years)", fontsize=22)
    ax.set_ylabel("Predicted age (years)", fontsize=22)
    ax.set_title(f"{name}\nMAE = {mae:.2f} yrs",
                 fontsize=20, fontweight="bold")
    ax.tick_params(axis="both", labelsize=15)
    ax.legend(fontsize=15)

plt.savefig("pred_vs_ca.png", dpi=150, bbox_inches="tight")
plt.show()
```

### BAG vs. chronological age

```python
# Create figure
fig, axes = plt.subplots(1, len(models), figsize=(6 * len(models), 5))
fig.suptitle("Brain age gap versus chronological age", fontsize=25, fontweight="bold", y=1.02)

# Create regression line
for ax, (name, df) in zip(axes, models.items()):
    slope_b, intercept_b, r_b, p_b, *_ = stats.linregress(df["Age_in_Yrs"], df["BAG"])

    sns.scatterplot(data=df, x="Age_in_Yrs", y="BAG", color=palette_models[name], alpha=0.8,
                    edgecolor="white", linewidth=0.5, s=70, ax=ax)
    ax.axhline(0, color="k", lw=1.2, linestyle="--", label="BAG = 0")
    ax.plot(x_line, intercept_b + slope_b * x_line, color="gray", lw=1.5,
            label=f"Regression")
    ax.set_xlim(age_min, age_max)
    ax.set_xlabel("Chronological age (years)")
    ax.set_ylabel("Brain age gap (years)")
    ax.set_title(f"{name}\n r = {r_b:.2f} | p = {p_b:.3f} ", fontsize=20, fontweight = "bold")
    ax.legend(fontsize=12)

plt.tight_layout()
plt.savefig("bag_vs_ca.png", dpi=150, bbox_inches="tight")
plt.show()
```
